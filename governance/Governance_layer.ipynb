```sql
%sql

CREATE CATALOG IF NOT EXISTS portfolio_catalog;

CREATE SCHEMA IF NOT EXISTS portfolio_catalog.bronze;
CREATE SCHEMA IF NOT EXISTS portfolio_catalog.silver;
CREATE SCHEMA IF NOT EXISTS portfolio_catalog.gold;
```

```sql
%sql
-- Bronze
CREATE TABLE portfolio_catalog.bronze.taxi_trips AS
SELECT * FROM portfolio.raw_data.bronze_taxi_trips;

-- Silver
CREATE TABLE portfolio_catalog.silver.taxi_trips AS
SELECT * FROM portfolio.raw_data.silver_taxi_trips;

-- Gold
CREATE TABLE portfolio_catalog.gold.daily_taxi_summary AS
SELECT * FROM portfolio.raw_data.gold_daily_taxi_summary;
```

```sql
%sql
-- Data Analysts: read-only on Gold
GRANT SELECT ON SCHEMA portfolio_catalog.gold TO `data_analysts`;

-- Data Engineers: full access Bronze to Gold
GRANT ALL PRIVILEGES ON CATALOG portfolio_catalog TO `data engineers`;

-- Explicitly block analysts from Bronze/Silver
REVOKE ALL PRIVILEGES ON SCHEMA portfolio_catalog.bronze FROM `data_analysts`;
REVOKE ALL PRIVILEGES ON SCHEMA portfolio_catalog.silver FROM `data_analysts`;
```

```sql
DROP VIEW IF EXISTS portfolio_catalog.silver.taxi_trips_masked;
```

```sql
CREATE VIEW portfolio_catalog.silver.taxi_trips_masked AS
SELECT
  VendorID,
  tpep_pickup_datetime,
  tpep_dropoff_datetime,
  passenger_count,
  trip_distance,
  CASE
    WHEN is_account_group_member('data_engineers') THEN PULocationID
    ELSE NULL
  END AS PULocationID,
  CASE
    WHEN is_account_group_member('data_engineers') THEN DOLocationID
    ELSE NULL
  END AS DOLocationID,
  payment_type,
  fare_amount,
  tip_amount,
  tolls_amount
FROM portfolio_catalog.silver.taxi_trips;
```

```sql
GRANT SELECT ON VIEW portfolio_catalog.silver.taxi_trips_masked TO `data_analysts`;
```

```sql
SELECT * FROM portfolio_catalog.silver.taxi_trips_masked LIMIT 10;
```

```sql
-- Step 1: Create a row filter function
-- Logic: data_engineers see everything; data_analysts see only VendorID = 1 (simulating a region/vendor restriction)
CREATE OR REPLACE FUNCTION portfolio_catalog.silver.vendor_row_filter(vendor_id INT)
RETURN is_account_group_member('data_engineers') OR vendor_id = 1;
```

```sql
-- Step 2: Apply the row filter to the Silver table
ALTER TABLE portfolio_catalog.silver.taxi_trips
SET ROW FILTER portfolio_catalog.silver.vendor_row_filter ON (VendorID);
```

```sql
-- Step 3: Test — count rows visible right now (as current user/group)
SELECT VendorID, COUNT(*) AS trip_count
FROM portfolio_catalog.silver.taxi_trips
GROUP BY VendorID;
```
